````python
import csv

with open("students.csv", "r", encoding="utf-8") as f:
    reader = csv.reader(f)
    columns = next(reader)
    rows = list(reader)

students = []

for row in rows:
    student = dict(zip(columns, row))
    subjects = columns[2:]
    student["grades"] = {s: int(student.pop(s)) for s in subjects}
    student["average_grade"] = round(
        sum(student["grades"].values()) / len(student["grades"]), 2
    )
    students.append(student)

print(students[0])
````

# Task 4: Replacing Words

````python
with open("words.txt", "r", encoding="utf-8") as f:
    text = f.read()

with open("forbidden_words.txt", "r", encoding="utf-8") as f:
    forbidden_words = f.read().split()

result = text

for word in forbidden_words:
    start = 0
    while True:
        index = result.lower().find(word.lower(), start)
        if index == -1:
            break
        result = result[:index] + "*" * len(word) + result[index + len(word):]
        start = index + len(word)

print(result)
````

# Answers to Self-Check Questions

### 1. How can you open a file for reading and writing?
Use `open()` with mode `'r+'`. It reads and writes without truncating, and the file must already exist.
```python
with open("file.txt", "r+", encoding="utf-8") as f:
    data = f.read()
    f.write("new text")
```
The other read-write modes are `'w+'`, which truncates or creates the file, and `'a+'`, which appends and can also read.

### 2. Difference between `'r'`, `'w'`, `'a'`, `'rb'`, `'wb'`
| Mode | Meaning |
|---|---|
| `'r'` | Read text (default). Raises `FileNotFoundError` if the file is missing |
| `'w'` | Write text. Creates the file or **erases** existing content |
| `'a'` | Append text to the end. Creates the file if missing |
| `'rb'` | Read binary. Returns `bytes` |
| `'wb'` | Write binary. Creates the file or erases it, and accepts `bytes` |

### 3. Advantage of `with open(...) as f:`
It is a context manager. The file closes automatically when the block ends, even if an exception is raised. This avoids resource leaks and the need to call `close()` manually.

### 4. How to close a file correctly, and why
Call `f.close()`, ideally in a `try/finally` block, or use `with`, which does this for you. Closing matters for three reasons:
- It flushes buffered data to disk. Without it, written data may be lost.
- It releases the OS file descriptor, of which there is a limited number.
- It unlocks the file for other processes, which is especially important on Windows.

### 5. Main `os` functions for directories
- `os.getcwd()` returns the current directory, and `os.chdir()` changes it.
- `os.listdir()` and `os.scandir()` list a directory's contents.
- `os.mkdir()` and `os.makedirs()` create directories.
- `os.rmdir()` and `os.removedirs()` remove empty directories.
- `os.rename()` renames a file or directory.
- `os.walk()` traverses a directory tree recursively.

### 6. Creating and deleting folders
```python
import os, shutil

os.mkdir("new_folder")                  # single folder
os.makedirs("a/b/c", exist_ok=True)     # nested folders
os.rmdir("new_folder")                  # only if empty
shutil.rmtree("a")                      # folder with all contents
```

### 7. `os.getcwd()` vs `os.chdir()`
- `os.getcwd()` **returns** the path of the current working directory.
- `os.chdir(path)` **changes** the current working directory to `path`.

### 8. Listing files and folders
```python
import os
from pathlib import Path

os.listdir(".")                            # list of names
[e.name for e in os.scandir(".")]          # entries with metadata
list(Path(".").iterdir())                  # Path objects
list(Path(".").glob("*.txt"))              # by pattern
```

### 9. What `os.path` is for
It handles path manipulation and checks:
- Joining paths: `join`
- Splitting paths: `split`, `splitext`, `basename`, `dirname`
- Checking existence and type: `exists`, `isfile`, `isdir`
- Converting to absolute paths: `abspath`
- Getting file size: `getsize`

### 10. Checking whether a file or folder exists
```python
import os
from pathlib import Path

os.path.exists("file.txt")   # file or folder
os.path.isfile("file.txt")
os.path.isdir("folder")

Path("file.txt").exists()
Path("folder").is_dir()
```

### 11. Why `pathlib` is more convenient than `os.path`
- It is object-oriented: a path is a `Path` object with methods, not a plain string.
- The `/` operator joins paths: `Path("data") / "file.txt"`.
- It has built-in helpers such as `.read_text()`, `.write_text()`, `.exists()`, `.mkdir()`, `.glob()`, `.suffix`, `.stem` and `.parent`.
- It is cross-platform by design, and the code is shorter and more readable.

### 12. Cross-platform path joining
```python
import os
from pathlib import Path

os.path.join("data", "files", "a.txt")
Path("data") / "files" / "a.txt"
```
Both use the correct separator for the OS: `\` on Windows and `/` on Linux and macOS.

### 13. Methods for reading a CSV file
```python
import csv

with open("data.csv", newline="", encoding="utf-8") as f:
    for row in csv.reader(f):          # rows as lists
        print(row)

with open("data.csv", newline="", encoding="utf-8") as f:
    for row in csv.DictReader(f):      # rows as dicts (header = keys)
        print(row["name"])
```
Another option is `pandas.read_csv("data.csv")`, which loads the file into a DataFrame.

### 14. Advantage of the `csv` module over plain line reading
A naive `line.split(",")` fails in several common cases. The `csv` module handles them correctly:
- Quoted fields containing commas, such as `"Astana, KZ"`
- Newlines inside quoted fields
- Escaped quotes
- Custom delimiters and dialects, such as `;` or `\t`

It also maps columns to header names through `DictReader` and `DictWriter`.

### 15. Serializing Python data to JSON
```python
import json

data = {"name": "Dias", "skills": ["Go", "Python"]}

s = json.dumps(data, ensure_ascii=False, indent=2)   # to string
with open("data.json", "w", encoding="utf-8") as f:
    json.dump(data, f, ensure_ascii=False, indent=2) # to file

obj = json.loads(s)                                  # from string
with open("data.json", encoding="utf-8") as f:
    obj = json.load(f)                               # from file
```

### 16. JSON vs CSV
| | JSON | CSV |
|---|---|---|
| Structure | Hierarchical: nested objects and lists | Flat table of rows and columns |
| Types | Strings, numbers, booleans, null | Everything is a string |
| Size | More verbose | Compact |
| Typical use | APIs, configs, complex data | Tabular data, Excel, data exports |

### 17. Python types that map directly to JSON
| Python | JSON |
|---|---|
| `dict` | object |
| `list`, `tuple` | array |
| `str` | string |
| `int`, `float` | number |
| `True` / `False` | true / false |
| `None` | null |

Types such as `set`, `datetime`, `bytes` and custom objects need conversion first, for example through `default=`. Dict keys become strings.

### 18. Lazy loading
Lazy loading means data is loaded or computed **only when it is actually needed**, not all at once. This has three benefits:
- Memory use stays low, so files larger than RAM can be processed.
- Processing can start sooner.
- No work is wasted on data that never gets used.

### 19. Reading a large file line by line
```python
with open("big.txt", encoding="utf-8") as f:
    for line in f:          # reads one line at a time
        process(line)
```
For binary data, read in chunks instead:
```python
with open("big.bin", "rb") as f:
    while chunk := f.read(8192):
        process(chunk)
```

### 20. `readline()` and iterating over a file object
- `f.readline()` reads **one line** per call and returns `""` at end of file. It is useful for reading a header or for controlled manual reading.
- `for line in f:` is the file's built-in lazy iterator. It is the idiomatic, memory-efficient way to process every line.
- For contrast, `f.readlines()` loads **all** lines into a list, so it is unsuitable for large files.

### 21. When to use generators with files
Use generators when:
- the file is large and should be processed as a stream;
- you are building a processing pipeline, such as read, then filter, then transform;
- you need only part of the data, such as the first N matches.

```python
def read_errors(path):
    with open(path, encoding="utf-8") as f:
        for line in f:
            if "ERROR" in line:
                yield line.strip()

for err in read_errors("app.log"):
    print(err)
```

### 22. Modules for archives
- `zipfile` handles `.zip`.
- `tarfile` handles `.tar`, `.tar.gz` and `.tar.bz2`.
- `shutil` offers high-level helpers: `make_archive()` and `unpack_archive()`.
- `gzip`, `bz2` and `lzma` compress single files.

### 23. Creating a ZIP archive
```python
import zipfile, shutil

with zipfile.ZipFile("archive.zip", "w", zipfile.ZIP_DEFLATED) as z:
    z.write("file1.txt")
    z.write("data/file2.csv", arcname="file2.csv")

shutil.make_archive("backup", "zip", "folder_to_archive")  # whole folder
```

### 24. Extracting an archive to a specified directory
```python
import zipfile, tarfile, shutil

with zipfile.ZipFile("archive.zip") as z:
    z.extractall("output_dir")

with tarfile.open("archive.tar.gz") as t:
    t.extractall("output_dir", filter="data")   # safe filter (Python 3.12+)

shutil.unpack_archive("archive.zip", "output_dir")
```
Be careful with archives from untrusted sources, because they can contain paths like `../` that write outside the target directory.

### 25. Key considerations for binary files
- Open them in binary mode (`'rb'` or `'wb'`). Otherwise Python decodes or encodes the content and corrupts it.
- Don't pass `encoding=`. You work with `bytes`, not `str`.
- Read large files in chunks rather than all at once.
- Byte structure matters. Use `struct` to parse formats and check magic numbers such as the PNG signature `b'\x89PNG'`.
- Use `seek()` and `tell()` for random access.
- Copy files with `shutil.copyfile()` instead of reading and writing them manually.
